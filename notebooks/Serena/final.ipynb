Missing Time Steps

The Time column has mixed formats, some rows including fractional seconds and some not. Here's how the timestamps are stored in our network logs. Instead of clean 1-second steps, the 'Time' column records events with microsecond precision (fractions of a second). Our network logs don't have a fixed interval, but rather are event-driven.

Our physical logs instead show that every consecutive timestamp is separated by 1 second with only one unique interval. Our network logs are packet-driven, and our physical logs are time-driven, sampled once per second.

```python
# Flag missing time steps across physical files
physical_files = [
    "phy_att_1.csv", "phy_att_2.csv", "phy_att_3.csv",
    "phy_att_4.csv", "phy_norm.csv"
]

for f in physical_files:
    try:
        # Load with UTF-16 encoding and tab delimiter for all except phy_att_4.csv
        if f == "phy_att_4.csv":
          df = pd.read_csv(physical_path + f, sep=",", encoding="utf-8")
          df['Time'] = pd.to_datetime(df['Time'], dayfirst=True, errors='coerce')
        else:
          df = pd.read_csv(physical_path + f, sep="\t", encoding="utf-16")
          df['Time'] = pd.to_datetime(df['Time'], dayfirst=True, errors='coerce')

        # Calculate differences between consecutive timestamps
        diffs = df['Time'].diff().dropna()

        # Count how many gaps are larger than 1 second
        missing_steps = (diffs > pd.Timedelta(seconds=1)).sum()

        print(f"\n{f}")
        print("Rows:", df.shape[0])
        print("Unique intervals:", diffs.unique())
        print("Missing steps:", missing_steps)

    except Exception as e:
        print(f"Error reading {f}: {e}")
```

Resample the network logs to 1-second bins to get packet counts per second. Use this to align network and physical logs perfectly.

Issue detected:
1. Our network timestamps use YYYY-MM-DD, while our physical timestamps use YYYY-DD-MM.

```python
# Force Pandas to interpret the physical timestamps as day-first
physical_df['time'] = pd.to_datetime(
    physical_df['time'],
    dayfirst=True,
    errors='coerce'
)

# Force Pandas to interpret normal.csv as NOT day-first
network_df['time'] = pd.to_datetime(
    network_df['time'],
    dayfirst=False,
    errors='coerce'
)

# Set the indexes
physical_df = physical_df.set_index('time').sort_index()
network_df = network_df.set_index('time').sort_index()
```

```python
# Verify the network file and physical file timestamps have the same format
print(network_df.index.min(), network_df.index.max())
print(physical_df.index.min(), physical_df.index.max())
```

```python
# Resample network logs to 1-second bins
network_1s = network_df.resample('1s').size().rename("packet_count")
```

*Now, we have acquired timestamp --> number of packets in that second.* Physical logs are sampled at 1s already, so just ensure the timestamp is parsed and set as the index.

### 4.3 - Data Selection

To detect anomalies, we'll use:
1. All physical sensors (tank levels, flow sensors, pressure sensors, pump states, valve states)
2. All network features (packet_count, rolling packet rates, packet deltas, network z-scores)
3. Derived features (flow/tank gradients, actuator state changes, rolling std for noise, rolling means for smoothing)

Now that our aligned dataframe has raw physical sensor values, raw network packet counts and labels, we can engineer features to capture behavior over time. We need to consider that ICS processes may change slowly, so we need derivatives and rolling windows.

```python
# Create a binary anomaly label
full_aligned_dataset['is_anomaly'] = (full_aligned_dataset['label'] != 'normal').astype(int)
```

```python
# Encode attack types
full_aligned_dataset['attack_type'] = full_aligned_dataset['label'].where(
    full_aligned_dataset['label'] != 'normal', 'none'
)
```

We need to see packet bursts during attacks, physical disturbances and timing relationships. For building a classifier, we need to drop non-numeric columns, fill NaNs, scale features and conduct a train/test split.

**Derived features**

```python
# Tank levels, packet_count, actuator deltas get rolling features
rolling_cols = [
    'tank_1','tank_2','tank_3','tank_4','tank_5','tank_6','tank_7','tank_8',
    'packet_count'
]
```

```python
# Sort by scenario + time
full_aligned_dataset = full_aligned_dataset.sort_values(
    ['scenario', full_aligned_dataset.index.name]
)
```

```python
# Rolling mean per scenario
window = 5  # 5-second rolling window

full_aligned_dataset[[f"{c}_roll_mean" for c in rolling_cols]] = (
    full_aligned_dataset.groupby('scenario')[rolling_cols]
        .rolling(window=window, min_periods=1)
        .mean()
        .reset_index(level=0, drop=True)
)
```

```python
# Rolling std per scenario
full_aligned_dataset[[f"{c}_roll_std" for c in rolling_cols]] = (
    full_aligned_dataset.groupby('scenario')[rolling_cols]
        .rolling(window=window, min_periods=1)
        .std()
        .reset_index(level=0, drop=True)
)
```

```python
# Rolling deltas
for c in rolling_cols:
    full_aligned_dataset[f"{c}_delta"] = (
        full_aligned_dataset.groupby('scenario')[c].diff()
    )
```

```python
# Rolling z-scores
for c in rolling_cols:
    full_aligned_dataset[f"{c}_z"] = (
        (full_aligned_dataset[c] - full_aligned_dataset[f"{c}_roll_mean"]) /
        (full_aligned_dataset[f"{c}_roll_std"] + 1e-6)
    )
```

Now are data is fully-aligned, cleaned, merged across all scenarios and has been enriched with rolling features to capture both short-term and long-term trends in our CIS. It can be used for anomaly detection.

### 4.4 - Handling Outliers

Models:
1. Regression - remove/winsorize
2. KNN - scale + remove
3. Naive Bayes - extreme values removed
4. SVM - scale + remove
5. DecisionTree - keep
6. Autoencoders = keep
7. GNN - imputation + scale + remove

```python
# Identify outliers via z-scores
outlier_mask = full_aligned_dataset[[c + "_z" for c in rolling_cols]].abs() > 5
```

This flags values more than 5 standard deviations above normal behavior.

```python
# Cap outliers instead of removing rows (winsorization)
for c in rolling_cols:
    full_aligned_dataset[c] = full_aligned_dataset[c].clip(
        lower=full_aligned_dataset[c].quantile(0.001),
        upper=full_aligned_dataset[c].quantile(0.999)
    )
```

```python
# Use a robust estimate as the absolute tank max
tank_cols = [f"tank_{i}" for i in range(1, 9)]
tank_max_est = full_aligned_dataset[tank_cols].quantile(0.999)
print(tank_max_est)
```

```python
# Fix impossible physical values
for t in tank_cols:
    full_aligned_dataset[t] = full_aligned_dataset[t].clip(
        lower=0,
        upper=tank_max_est[t]
    )
```

```python
# Fix actuator glitches (they should not toggle 10 times per second)
bool_cols = full_aligned_dataset.select_dtypes(bool).columns

for col in bool_cols:
    full_aligned_dataset[col] = (
        full_aligned_dataset.groupby("scenario")[col]
        .rolling(window=3, min_periods=1)
        .median()
        .reset_index(level=0, drop=True)
        .astype(bool)
    )
```

```python
# Fix packet_count spikes
full_aligned_dataset['packet_count'] = full_aligned_dataset['packet_count'].clip(
    upper=full_aligned_dataset['packet_count'].quantile(0.999)
)
```

```python
# Normalize tank values
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
full_aligned_dataset[tank_cols] = scaler.fit_transform(full_aligned_dataset[tank_cols])
```

Our tank values are clean and normalized, but we do have outliers only in network features.

```python
# Network features to normalize
network_cols = [
    'packet_count', 'packet_delta', 'packet_count_delta',
    'packet_rate_5s', 'packet_rate_30s', 'packet_std_30s',
    'pkt_mean_5s', 'pkt_mean_30s', 'pkt_std_30s',
    'pkt_rate_change', 'packet_count_roll_mean',
    'packet_count_roll_std', 'packet_count_z'
]

flow_cols = ['flow_sensor_1', 'flow_sensor_2', 'flow_sensor_3', 'flow_sensor_4']
```

```python
# Clip outliers using 0.1% - 99.9% percentiles
for col in network_cols + flow_cols:
    lower = full_aligned_dataset[col].quantile(0.001)
    upper = full_aligned_dataset[col].quantile(0.999)
    full_aligned_dataset[col] = full_aligned_dataset[col].clip(lower, upper)
```

This removes packet bursts, flow sensor spikes, delta explosions and z-score anomalies without removing rows.
