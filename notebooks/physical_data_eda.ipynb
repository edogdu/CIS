# Physical Data Analysis

## Which valves are actually used?
There are some valves that are in Figure 2 (Real vs Simulated partition Diagram) that are excluded from Figure 3 (Physical Process Diagram)

```python
import pandas as pd
import os


phys_data_path = '../data/testbed_system_1/Physical'
cols = []
for f in os.listdir(phys_data_path):
    df = None    
    if f.endswith('phy_att_4.csv'):
        df = pd.read_csv(os.path.join(phys_data_path, f))
    else:
        df = pd.read_csv(os.path.join(phys_data_path, f), sep='\t', encoding='utf-16')
    df_filtered = df.filter(regex='Valv')
    df_filtered = df_filtered[df_filtered.columns[df_filtered.nunique() != 1]]
    for col in df_filtered.columns:
        if col not in cols:
            cols.append(col)


print(f"Valve columns that have a value change in the dataset:\r\n{cols}")
```

### Results
The Valves that have a value change is consistent with Physical Process Diagram.
